```python
import pandas as pd

df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

print("=== 5 Baris Pertama Data ===")
display(df.head())


df['kode_kasir'] = df['kode_kasir'].fillna('K_UNKNOWN')
df['diskon_persen'] = df['diskon_persen'].fillna(0)


df['harga_satuan_clean'] = df['harga_satuan'].astype(str).str.replace(',', '.').astype(float)
df['tanggal_clean'] = pd.to_datetime(df['tanggal'], format='mixed', dayfirst=True, errors='coerce')
df['tahun'] = df['tanggal_clean'].dt.year

df['pendapatan'] = df['jumlah'] * df['harga_satuan_clean'] * (1 - df['diskon_persen'] / 100.0)

df_selesai = df[df['status'] == 'SELESAI'].copy()

df_2024 = df_selesai[df_selesai['tahun'] == 2024].copy()
df_2025 = df_selesai[df_selesai['tahun'] == 2025].copy()

df_2024.to_csv('transaksi_2024.csv', index=False)
df_2025.to_csv('transaksi_2025.csv', index=False)

print("\n" + "="*50)
print("HASIL ANALISIS PENDAPATAN PER TAHUN")
print("="*50)
print(f"Tahun 2024:")
print(f"  - Jumlah Data  : {len(df_2024):,} transaksi")
print(f"  - Pendapatan   : Rp {df_2024['pendapatan'].sum():,.2f}")
print("-"*50)
print(f"Tahun 2025:")
print(f"  - Jumlah Data  : {len(df_2025):,} transaksi")
print(f"  - Pendapatan   : Rp {df_2025['pendapatan'].sum():,.2f}")
print("="*50)
```
